```python
from sklearn.decomposition import PCA
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
X = df[['Age','Nationality']]
```

```python
X = X.fillna(X.mean())
```

```python
X_scaled = StandardScaler().fit_transform(X)
```

```python
X_cleaned = X.dropna(axis=1)
X_scaled_clean = StandardScaler().fit_transform(X_cleaned)
pca = PCA(n_components=1)
principal_components = pca.fit_transform(X_scaled_clean)
print("PCA computed successfully. Shape:", principal_components.shape)
```

```python
import matplotlib.pyplot as plt

# Since there is only 1 Principal Component, we can plot it against its index
plt.scatter(range(len(principal_components)), principal_components[:, 0], alpha=0.5, s=1)
plt.xlabel('Sample Index')
plt.ylabel('Principal Component 1')
plt.title('1D PCA Visualization (Projected Age Feature)')
plt.show()
```
